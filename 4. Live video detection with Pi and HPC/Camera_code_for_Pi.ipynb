# Code to run on the Pi to create an RTSP stream at 10fps

```python
#!/bin/bash

sudo -v && sudo rpicam-vid \
  --width 1920 --height 1080 \
  --roi 0.2879,0.2557,0.4685,0.5866 \
  --framerate 10 \
  --codec mjpeg \
  -t 0 -n -o - | \
ffmpeg -f mjpeg -i - -c copy -f rtsp -rtsp_transport tcp rtsp://localhost:8554/stream1
```
